## 1. Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)

import warnings
warnings.filterwarnings('ignore')
```

## 2. Load Raw Dataset

```python
raw = pd.read_csv(r"C:\Users\admin\Downloads\Teen_Health_Risk_Dataset.csv")
df = raw.copy()

df.head()
```

## 3. Understand the Dataset

```python
# Shape
print(df.shape)
# Column names
print(df.columns)
# Data types
print(df.dtypes)
# Dataset information
print(df.info())
# Numerical summary
print(df.describe())
# Categorical summary
print(df.describe(include='object'))
```

## 4. Missing Values

```python
df.isnull().sum()

missing = (df.isnull().sum() / len(df)) * 100
missing.sort_values(ascending=False)
```

```python
# Fill missing categorical values with meaningful categories
num_cols = df.select_dtypes(include=['number']).columns
cat_cols = df.select_dtypes(include=['object']).columns

df['Medication_Use'].fillna('No Medictaion', inplace=True)
df['Existing_Conditions'] = df['Existing_Conditions'].fillna('No Known Condition')
df['Alcohol_Consumption'] = df['Alcohol_Consumption'].fillna('Non-Drinker')
df['Whole_Grain_Intake'] = df['Whole_Grain_Intake'].fillna('Unknown')
```

## 5. Duplicate Records

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
df.shape
```

## 6. Unique Values and Text Cleaning

```python
for col in cat_cols:
    print(col)
    print(df[col].unique())
    print("-" * 50)
```

```python
# Remove extra spaces from categorical columns
for col in cat_cols:
    df[col] = df[col].str.strip()
```

## 7. Outlier and Error Checks

```python
num_cols = df.select_dtypes(include=['int64', 'float64']).columns

plt.figure(figsize=(20, 15))

for i, col in enumerate(num_cols):
    plt.subplot(7, 6, i + 1)
    sns.boxplot(y=df[col])

plt.tight_layout()
plt.show()
```

```python
# IQR-based inspection only; extreme values are reviewed rather than blindly removed.
num_cols = df.select_dtypes(include="number")

for col in num_cols:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1
    lb = q1 - 1.5 * iqr
    ub = q3 + 1.5 * iqr
    outliers = df[(df[col] < lb) | (df[col] > ub)]

    print(col)
    print("outlier count -->", len(outliers))
    print("upper_bound -->", ub)
    print("lower_bound -->", lb)
    print()
```

## 8. Handle Implausible Values

```python
df['Height_cm'] = df['Height_cm'].clip(130, 220)
df['Weight_kg'] = df['Weight_kg'].clip(25, 180)
df['Water_Intake_Liters'] = df['Water_Intake_Liters'].clip(0.5, 12)
df['Sleep_Hours'] = df['Sleep_Hours'].clip(3, 15)
df['Screen_Time_Hours_Per_Day'] = df['Screen_Time_Hours_Per_Day'].clip(0, 24)
df['Exercise_Frequency_Per_Week'] = df['Exercise_Frequency_Per_Week'].clip(0, 7)
df['Fruit_Servings_Per_Day'] = df['Fruit_Servings_Per_Day'].clip(0, 10)
df['Vegetable_Servings_Per_Day'] = df['Vegetable_Servings_Per_Day'].clip(0, 10)
df['Saturated_Fat_g'] = df['Saturated_Fat_g'].clip(0, 90)
df['Protein_g'] = df['Protein_g'].clip(0, 200)
```
